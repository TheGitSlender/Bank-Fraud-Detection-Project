```python
import pandas as pd
import numpy as np
```

```python
train_identity = pd.read_csv("data/train_identity.csv")
```

```python
train_transaction = pd.read_csv("data/train_transaction.csv")
train_transaction = train_transaction.drop(columns= [col for col in train_transaction.columns if col.startswith('V')], axis = 1)
```

```python
null_values = train_transaction.isnull().sum()/len(train_transaction)
dropped_half_df = train_transaction.drop(columns=[col for col in train_transaction.columns if null_values[col] > 0.5], axis=1)
dropped_eight_df = train_transaction.drop(columns=[col for col in train_transaction.columns if null_values[col] > 0.8], axis=1)
```

```python
for col in dropped_eight_df.select_dtypes(exclude=["object"]).columns:
    dropped_eight_df[col].fillna(dropped_eight_df[col].mean(), inplace=True)
```

```python
# Basic dataset structure
print("=" * 80)
print("DATASET STRUCTURE OVERVIEW")
print("=" * 80)
print(f"\nTrain Transaction Shape: {train_transaction.shape}")
print(f"Train Identity Shape: {train_identity.shape}")
print(f"\nTransaction Columns: {train_transaction.shape[1]}")
print(f"Identity Columns: {train_identity.shape[1]}")
print(f"\nTotal Transactions: {train_transaction.shape[0]:,}")
print(f"Transactions with Identity Info: {train_identity.shape[0]:,}")
print(f"Percentage with Identity: {(train_identity.shape[0] / train_transaction.shape[0] * 100):.2f}%")
```

```python
# Data types analysis
print("\n" + "=" * 80)
print("DATA TYPES BREAKDOWN")
print("=" * 80)

print("\nTRANSACTION DATA:")
print(f"  Numeric (float64): {(train_transaction.dtypes == 'float64').sum()}")
print(f"  Numeric (int64): {(train_transaction.dtypes == 'int64').sum()}")
print(f"  Categorical (object): {(train_transaction.dtypes == 'object').sum()}")
print(f"  Memory Usage: {train_transaction.memory_usage(deep=True).sum() / 1024**3:.2f} GB")

print("\nIDENTITY DATA:")
print(f"  Numeric (float64): {(train_identity.dtypes == 'float64').sum()}")
print(f"  Numeric (int64): {(train_identity.dtypes == 'int64').sum()}")
print(f"  Categorical (object): {(train_identity.dtypes == 'object').sum()}")
print(f"  Memory Usage: {train_identity.memory_usage(deep=True).sum() / 1024**2:.2f} MB")

# List categorical columns
print("\nCATEGORICAL COLUMNS IN TRANSACTION:")
cat_cols_trans = train_transaction.select_dtypes(include='object').columns.tolist()
print(cat_cols_trans)

print("\nCATEGORICAL COLUMNS IN IDENTITY:")
cat_cols_identity = train_identity.select_dtypes(include='object').columns.tolist()
print(cat_cols_identity)
```

```python
# Target variable analysis
print("\n" + "=" * 80)
print("TARGET VARIABLE ANALYSIS (isFraud)")
print("=" * 80)

fraud_counts = train_transaction['isFraud'].value_counts()
fraud_pct = train_transaction['isFraud'].value_counts(normalize=True) * 100

print(f"\nFraud Distribution:")
print(f"  Non-Fraud (0): {fraud_counts[0]:,} ({fraud_pct[0]:.2f}%)")
print(f"  Fraud (1): {fraud_counts[1]:,} ({fraud_pct[1]:.2f}%)")
print(f"\nClass Imbalance Ratio: {fraud_counts[0] / fraud_counts[1]:.2f}:1")
print(f"Total Fraudulent Transactions: {fraud_counts[1]:,}")
print(f"Fraud Rate: {fraud_pct[1]:.2f}%")

# Check fraud rate in identity subset
merged = train_transaction[['TransactionID', 'isFraud']].merge(
    train_identity[['TransactionID']], 
    on='TransactionID', 
    how='inner'
)
print(f"\nFraud Rate in Transactions WITH Identity Info: {merged['isFraud'].mean() * 100:.2f}%")
print(f"Fraud Rate in Transactions WITHOUT Identity Info: {train_transaction[~train_transaction['TransactionID'].isin(train_identity['TransactionID'])]['isFraud'].mean() * 100:.2f}%")
```

```python
# Missing values analysis
print("\n" + "=" * 80)
print("MISSING VALUES ANALYSIS")
print("=" * 80)

# Transaction data
trans_missing = train_transaction.isnull().sum()
trans_missing_pct = (trans_missing / len(train_transaction) * 100).sort_values(ascending=False)

print("\nTRANSACTION DATA:")
print(f"  Columns with NO missing values: {(trans_missing == 0).sum()}")
print(f"  Columns with missing values: {(trans_missing > 0).sum()}")
print(f"  Columns with >50% missing: {(trans_missing_pct > 50).sum()}")
print(f"  Columns with >90% missing: {(trans_missing_pct > 90).sum()}")

print("\nTop 10 columns with highest missing percentage:")
for col, pct in trans_missing_pct.head(10).items():
    print(f"  {col}: {pct:.2f}%")

# Identity data
identity_missing = train_identity.isnull().sum()
identity_missing_pct = (identity_missing / len(train_identity) * 100).sort_values(ascending=False)

print("\nIDENTITY DATA:")
print(f"  Columns with NO missing values: {(identity_missing == 0).sum()}")
print(f"  Columns with missing values: {(identity_missing > 0).sum()}")
print(f"  Columns with >50% missing: {(identity_missing_pct > 50).sum()}")
print(f"  Columns with >90% missing: {(identity_missing_pct > 90).sum()}")

print("\nTop 10 columns with highest missing percentage:")
for col, pct in identity_missing_pct.head(10).items():
    print(f"  {col}: {pct:.2f}%")
```

```python
# Key transaction features analysis
print("\n" + "=" * 80)
print("KEY TRANSACTION FEATURES ANALYSIS")
print("=" * 80)

# Transaction Amount
print("\nTRANSACTION AMOUNT:")
print(f"  Mean: ${train_transaction['TransactionAmt'].mean():.2f}")
print(f"  Median: ${train_transaction['TransactionAmt'].median():.2f}")
print(f"  Std Dev: ${train_transaction['TransactionAmt'].std():.2f}")
print(f"  Min: ${train_transaction['TransactionAmt'].min():.2f}")
print(f"  Max: ${train_transaction['TransactionAmt'].max():.2f}")

# By fraud status
fraud_amt = train_transaction[train_transaction['isFraud'] == 1]['TransactionAmt']
non_fraud_amt = train_transaction[train_transaction['isFraud'] == 0]['TransactionAmt']

print(f"\n  FRAUD Transactions:")
print(f"    Mean: ${fraud_amt.mean():.2f}")
print(f"    Median: ${fraud_amt.median():.2f}")
print(f"\n  NON-FRAUD Transactions:")
print(f"    Mean: ${non_fraud_amt.mean():.2f}")
print(f"    Median: ${non_fraud_amt.median():.2f}")

# TransactionDT (time)
print("\nTRANSACTION TIME (TransactionDT):")
print(f"  Time Range: {train_transaction['TransactionDT'].min():,.0f} to {train_transaction['TransactionDT'].max():,.0f}")
print(f"  Duration: ~{(train_transaction['TransactionDT'].max() - train_transaction['TransactionDT'].min()) / (60*60*24):.0f} days")

# Product categories
print("\nPRODUCT CATEGORIES (ProductCD):")
print(train_transaction['ProductCD'].value_counts())

# Card features
print("\nCARD FEATURES:")
print(f"  Unique card1 values: {train_transaction['card1'].nunique():,}")
print(f"  Unique card2 values: {train_transaction['card2'].nunique():,}")
print(f"  Unique card3 values: {train_transaction['card3'].nunique():,}")
print(f"  card4 (card type) unique values: {train_transaction['card4'].nunique()}")
print(f"  card6 (card category) unique values: {train_transaction['card6'].nunique()}")
```

```python
# Identity features analysis
print("\n" + "=" * 80)
print("IDENTITY FEATURES ANALYSIS")
print("=" * 80)

# Device information
print("\nDEVICE INFORMATION:")
print(f"  DeviceType unique values: {train_identity['DeviceType'].nunique()}")
print("\nDeviceType distribution:")
print(train_identity['DeviceType'].value_counts())

print(f"\n  DeviceInfo unique values: {train_identity['DeviceInfo'].nunique()}")
print("\nTop 10 DeviceInfo:")
print(train_identity['DeviceInfo'].value_counts().head(10))

# ID features statistics
print("\nID FEATURES (numeric):")
numeric_id_cols = [col for col in train_identity.columns if col.startswith('id_') and train_identity[col].dtype in ['float64', 'int64']]
for col in sorted(numeric_id_cols)[:5]:  # First 5
    print(f"\n  {col}:")
    print(f"    Missing: {train_identity[col].isnull().sum() / len(train_identity) * 100:.2f}%")
    if train_identity[col].notna().sum() > 0:
        print(f"    Mean: {train_identity[col].mean():.2f}")
        print(f"    Unique values: {train_identity[col].nunique()}")

# ID categorical features
print("\nID FEATURES (categorical):")
cat_id_cols = [col for col in train_identity.columns if col.startswith('id_') and train_identity[col].dtype == 'object']
for col in sorted(cat_id_cols)[:5]:  # First 5
    print(f"\n  {col}:")
    print(f"    Missing: {train_identity[col].isnull().sum() / len(train_identity) * 100:.2f}%")
    print(f"    Unique values: {train_identity[col].nunique()}")
    if train_identity[col].nunique() <= 10:
        print(f"    Values: {train_identity[col].value_counts().head().to_dict()}")
```

```python
# Categorical features analysis
print("\n" + "=" * 80)
print("CATEGORICAL FEATURES DETAILED ANALYSIS")
print("=" * 80)

# Email domains
print("\nEMAIL DOMAINS:")
print(f"\nP_emaildomain (Purchaser email):")
print(f"  Unique domains: {train_transaction['P_emaildomain'].nunique()}")
print(f"  Missing: {train_transaction['P_emaildomain'].isnull().sum() / len(train_transaction) * 100:.2f}%")
print("  Top 10 domains:")
print(train_transaction['P_emaildomain'].value_counts().head(10))

print(f"\nR_emaildomain (Recipient email):")
print(f"  Unique domains: {train_transaction['R_emaildomain'].nunique()}")
print(f"  Missing: {train_transaction['R_emaildomain'].isnull().sum() / len(train_transaction) * 100:.2f}%")
print("  Top 10 domains:")
print(train_transaction['R_emaildomain'].value_counts().head(10))

# M columns (match features)
print("\nMATCH FEATURES (M1-M9):")
for col in ['M1', 'M2', 'M3', 'M4']:
    print(f"\n{col}:")
    print(f"  Unique values: {train_transaction[col].nunique()}")
    print(f"  Missing: {train_transaction[col].isnull().sum() / len(train_transaction) * 100:.2f}%")
    if train_transaction[col].nunique() <= 5:
        print(f"  Distribution: {train_transaction[col].value_counts().to_dict()}")

# Card type and category
print("\nCARD TYPE (card4):")
print(train_transaction['card4'].value_counts())

print("\nCARD CATEGORY (card6):")
print(train_transaction['card6'].value_counts())
```

```python
# Additional insights and V features
print("\n" + "=" * 80)
print("V FEATURES (Vesta Engineered Features)")
print("=" * 80)

# Count V columns by prefix
v_cols = [col for col in train_transaction.columns if col.startswith('V')]
print(f"\nTotal V features: {len(v_cols)}")

# Group by ranges
v_ranges = {
    'V1-V11': [col for col in v_cols if int(col[1:]) <= 11],
    'V12-V34': [col for col in v_cols if 12 <= int(col[1:]) <= 34],
    'V35-V52': [col for col in v_cols if 35 <= int(col[1:]) <= 52],
    'V53-V74': [col for col in v_cols if 53 <= int(col[1:]) <= 74],
    'V75-V94': [col for col in v_cols if 75 <= int(col[1:]) <= 94],
    'V95-V137': [col for col in v_cols if 95 <= int(col[1:]) <= 137],
    'V138-V166': [col for col in v_cols if 138 <= int(col[1:]) <= 166],
    'V167-V216': [col for col in v_cols if 167 <= int(col[1:]) <= 216],
    'V217-V278': [col for col in v_cols if 217 <= int(col[1:]) <= 278],
    'V279-V321': [col for col in v_cols if 279 <= int(col[1:]) <= 321],
    'V322-V339': [col for col in v_cols if 322 <= int(col[1:]) <= 339],
}

print("\nV features grouped by ranges:")
for range_name, cols in v_ranges.items():
    missing_pct = train_transaction[cols].isnull().sum().sum() / (len(train_transaction) * len(cols)) * 100
    print(f"  {range_name}: {len(cols)} features, Avg missing: {missing_pct:.2f}%")

# D features (timedelta)
d_cols = [col for col in train_transaction.columns if col.startswith('D')]
print(f"\nD FEATURES (Timedelta features): {len(d_cols)}")
print("Missing values by D column:")
for col in d_cols:
    missing_pct = train_transaction[col].isnull().sum() / len(train_transaction) * 100
    print(f"  {col}: {missing_pct:.2f}%")

# C features (counting features)
c_cols = [col for col in train_transaction.columns if col.startswith('C')]
print(f"\nC FEATURES (Count features): {len(c_cols)}")
print(f"Overall missing rate: {train_transaction[c_cols].isnull().sum().sum() / (len(train_transaction) * len(c_cols)) * 100:.2f}%")
```
